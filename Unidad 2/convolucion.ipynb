```python
from PIL import Image
gato=Image.open("./data/images_dataset/gato/gato1.jpg")
gato
```

```python
from torchvision import transforms
import torch
import numpy as np
print(np.asarray(gato).shape)
transform = transforms.Compose([
    transforms.ToTensor()
])
img=torch.unsqueeze(transform(gato),0)
print(img.shape)
```

```python
conv=torch.nn.Conv2d(in_channels=3, out_channels=1, kernel_size=3, stride=1, padding=0)
img2=conv(img)
print(img2.shape)
```

```python
img3=img2.squeeze(0).permute(1,2,0).detach().numpy()
img3=img3/img3.max()
```

```python
import matplotlib.pyplot as plt
plt.imshow(img3, cmap='gray')
plt.show()
```

```python
# filtro
for i in range(3):
    x=conv.weight[0,i,:,:].detach().numpy()
    plt.subplot(1,3,i+1)
    plt.imshow(x, cmap='gray')
    plt.colorbar()
```

```python
conv=torch.nn.Conv2d(in_channels=3, out_channels=16, kernel_size=3, stride=1, padding=0)
img2=conv(img)
plt.figure(figsize=(16,16))
print(img.shape, img2.shape)
for i in range(16):
    x=img2[0,i,:,:].detach().numpy()
    x=x/x.max()
    plt.subplot(4,4,i+1)
    plt.imshow(x, cmap='gray')
```
