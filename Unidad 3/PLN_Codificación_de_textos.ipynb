# Codificación de textos

Vamos a comprender el problema de la codificación de los archivos de texto plano.

Cuando un archivo se abre, aparecen seguramente errores en la lectura del archivo o en la representación de ciertos caracteres especiales como las tildes, acentos y la ñ entre otros.

Vamos a considerar dos archivos y los abriremos con las opciones más comunes para detectar la codificación del mismo.


```python
# Intentamos leer el libro de los muertos por medio de una apertura del mismo sin parámetros
# La lectura se hace por defecto con el tipo utf-8, pero esto genera errores de interpretación de los datos. Entonces, lo más simple es ensayar otro tipo de decodificación.
with open('libro_muertos.txt') as file:
  lines=file.readlines()
#Nota, esta celda arrojará un error, no se preocupe por eso
```

```python
# El error: UnicodeDecodeError: 'utf-8' codec can't decode byte 0xed in position 225: invalid continuation byte
# Indica que el archivo no es utf-8. Debe abrirse con otra codificación
# Normalmente los archivos disponibles en español son utf o latin1.

with open('libro_muertos.txt', encoding='latin1') as file:
  lines=file.readlines()
for li in lines[0:20]:
  print(li)
```

```python
# El libro de la constitución si es un libro UTF-8, que puede ser abierto con la codificación por defecto
with open('constitucion.txt', encoding='utf-8') as file:
  lines=file.readlines()
for li in lines[0:10]:
  print(li)
```

UTF-8 y Latin-1 (ISO-8859-1) son dos codificaciones de caracteres: reglas que determinan cómo se convierten los caracteres de un texto en bytes para almacenarlos en un archivo. Latin-1 utiliza un byte por carácter y dispone de 256 valores posibles. Incluye las letras del alfabeto latino y caracteres habituales del español como á, é, í, ó, ú, ü y ñ, pero no puede representar muchos otros alfabetos ni símbolos modernos.

UTF-8 es una codificación de Unicode, el estándar que asigna un número único (code point) a caracteres de prácticamente todos los sistemas de escritura. UTF-8 representa esos caracteres mediante uno a cuatro bytes. Los caracteres ASCII (A-Z, a-z, números, etc.) ocupan un byte, mientras que caracteres como ñ o á normalmente ocupan dos. Por ejemplo, Unicode asigna a ñ el código U+00F1; en UTF-8 se almacena mediante los bytes C3 B1.

Actualmente, UTF-8 es la opción recomendada y dominante. Un archivo en español puede estar perfectamente codificado en Latin-1, pero UTF-8 permite mezclar sin problemas español con caracteres como α, π, 中文, 日本語, €, → o 😊. Por eso, salvo que se trabaje con sistemas antiguos que exijan Latin-1, conviene utilizar UTF-8.
